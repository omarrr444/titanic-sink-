```python
train_data = pd.read_csv('/content/train.csv')
test_data = pd.read_csv('/content/test.csv')
```

```python
train_data.head()
```

```python
print(train_data.isnull().sum())
```

```python
train_data['Age'] = train_data['Age'].fillna(train_data['Age'].mean())
```

```python
train_data = train_data.drop(['Cabin'], axis=1)
```

```python
train_data['Embarked'] = train_data['Embarked'].fillna(train_data['Embarked'].mode()[0])
```

```python
print(train_data.isnull().sum())
```

```python
print(train_data['Survived'].value_counts())
```

```python
import seaborn as sns
sns.set()

sns.countplot(x='Survived', data=train_data)
```

```python
print(train_data['Sex'].value_counts())
```

```python
sns.countplot(x='Sex', data=train_data)
```

```python

plt.figure(figsize=(6,4))
sns.countplot(x='Sex', data=train_data)
plt.title('(1) Passenger Gender Distribution')
plt.show()
```

```python

plt.figure(figsize=(6,4))
sns.countplot(x='Sex', hue='Survived', data=train_data)
plt.title('(2) Survival Distribution by Gender')
plt.legend(title='Survived', labels=['No', 'Yes'])
plt.show()
```

```python

plt.figure(figsize=(6,4))
sns.countplot(x='Pclass', hue='Survived', data=train_data)
plt.title('(3) Survival Distribution by Passenger Class')
plt.legend(title='Survived', labels=['No', 'Yes'])
plt.xlabel('Passenger Class')
plt.show()
```

```python
train_data.replace({'Sex': {'male': 0, 'female': 1},
                    'Embarked': {'S': 0, 'C': 1, 'Q': 2}},
                   inplace=True)
```

```python
X_train = train_data.drop(columns=['PassengerId', 'Survived', 'Name', 'Ticket'], axis=1)
Y_train = train_data['Survived']
```

```python
X_train.head()
Y_train.head()
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
```

```python
Model = LogisticRegression()
```

```python
Model.fit(X_train,Y_train)
```

```python
X_train_prediction = Model.predict(X_train)
training_accuracy = accuracy_score(Y_train, X_train_prediction)
print( training_accuracy)
```

```python
from sklearn.metrics import classification_report
print(classification_report(Y_train,X_train_prediction))
```

```python
from sklearn.tree import DecisionTreeClassifier
Model_dt = DecisionTreeClassifier()
```

```python
Model_dt.fit(X_train,Y_train)
```

```python
X_train_prediction_dt = Model_dt.predict(X_train)
training_accuracy_dt = accuracy_score(Y_train, X_train_prediction_dt)
print(training_accuracy_dt)
```

```python
from sklearn.metrics import classification_report
print(classification_report(Y_train,X_train_prediction_dt))
```
